```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread('/content/RGB image.jpg')
```

```python
top,bottom,left,right=50,50,50,50
```

```python
border_constant = cv2.copyMakeBorder(img, top, bottom, left, right, cv2.BORDER_CONSTANT, value=[255, 0, 0])
plt.imshow(cv2.cvtColor(border_constant, cv2.COLOR_BGR2RGB))
plt.title('BORDER_CONSTANT')
plt.show()
```

```python
border_replicate = cv2.copyMakeBorder(img, top, bottom, left, right, cv2.BORDER_REPLICATE)
plt.imshow(cv2.cvtColor(border_replicate, cv2.COLOR_BGR2RGB))
plt.title('BORDER_REPLICATE')
plt.show()
```

```python
border_reflect = cv2.copyMakeBorder(img, top, bottom, left, right, cv2.BORDER_REFLECT)
plt.imshow(cv2.cvtColor(border_reflect, cv2.COLOR_BGR2RGB))
plt.title('BORDER_REFLECT')
plt.show()
```

```python
border_reflect_101 = cv2.copyMakeBorder(img, top, bottom, left, right, cv2.BORDER_REFLECT_101)
plt.imshow(cv2.cvtColor(border_reflect_101, cv2.COLOR_BGR2RGB))
plt.title('BORDER_REFLECT_101')
plt.show()
```

```python
border_wrap = cv2.copyMakeBorder(img, top, bottom, left, right, cv2.BORDER_WRAP)
plt.imshow(cv2.cvtColor(border_wrap, cv2.COLOR_BGR2RGB))
plt.title('BORDER_WRAP')
plt.show()
```
