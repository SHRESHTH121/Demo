```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread('/content/IP_sample_img_blackAndWhite.webp', cv2.IMREAD_GRAYSCALE)
print("Min:", np.min(img), "Max:", np.max(img))
fmin, fmax = np.min(img), np.max(img)
stretched = ((img - fmin) / (fmax - fmin) * 255).astype(np.uint8)
plt.subplot(1,2,1); plt.imshow(img, cmap='gray'); plt.title('Original'); plt.axis('on')
plt.subplot(1,2,2); plt.imshow(stretched, cmap='gray'); plt.title('Contrast Stretched'); plt.axis('on')
plt.show()
```
