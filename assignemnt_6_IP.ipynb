for Brighter


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
img = cv2.imread('/content/Screenshot 2026-05-02 002952.png', cv2.IMREAD_GRAYSCALE)


gamma = 0.5
img_float = img.astype(np.float32) / 255.0
power_law = np.power(img_float, gamma)

power_law = (power_law * 255).astype(np.uint8)

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.axis('on')

plt.subplot(1,2,2)
plt.imshow(power_law, cmap='gray')
plt.title('Gamma = 0.5')
plt.axis('on')

plt.show()
```
