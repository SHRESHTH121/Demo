```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread('/content/IP_sample_img.webp')

negative = 255 - img

side_by_side = np.hstack((img, negative))

cv2.imwrite('side_by_side.jpg', side_by_side)
cv2_imshow(side_by_side)
```
