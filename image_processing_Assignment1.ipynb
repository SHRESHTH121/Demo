```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
img=cv2.imread("/content/gray scale.jpeg",0)
```

```python
threshold=np.mean(img)
```

```python
ret,binary=cv2.threshold(img,threshold,255,cv2.THRESH_BINARY)
```

```python
plt.subplot(1,2,1)
plt.imshow(img,cmap="gray")
plt.title('Gray_Scale')
plt.subplot(1,2,2)
plt.title('Binary')
plt.imshow(binary,cmap="gray")
```
